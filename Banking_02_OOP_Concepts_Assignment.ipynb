```python
#Q31. Inherit perishableaccount from account for banking case 1; use sample reference ACC-1031 and explain the result.


class Account:

  def __init__(self,Account_Number,Account_Name,Balance):
    self.Account_Number = Account_Number
    self.Account_Name = Account_Name
    self.Balance = Balance

class Perishableaccount(Account):
  def __init__(self,Account_Number,Account_Name,Balance,Expiry_date):
    super().__init__(Account_Number,Account_Name,Balance)
    self.Expiry_date = Expiry_date

  def display(self):
    print("Account Number :",self.Account_Number)
    print("Account Name   :",self.Account_Name)
    print("Balance        :",self.Balance)
    print("Expiry Date    :",self.Expiry_date)

account = Perishableaccount("ACC-1031","Nafim",50000,"01-08-2008")

account.display()

## Inheritance allows a child class to reuse the properties and methods of a parent class.
## Here, PerishableAccount inherits from Account, reducing code duplication while adding an additional attribute called expiry_date.
```

```python
#Q32. Override a discount method for banking case 1; use sample reference ACC-1032 and explain the result.


class Account:

  def __init__(self,Account_Number,Balance):
    self.Account_Number = Account_Number
    self.Balance = Balance

  def Discount(self):
    return self.Balance * 0.05

class vipaccount(Account):
  def Discount(self):
    return self.Balance * 0.10

account = [vipaccount("ACC-1032",60000),Account("Acc-1032",5000)]

for i in account:
  print("Account_Number  :",i.Account_Number)
  print('Balance         :',i.Balance)
  print("Discount        :",i.Discount())
  print()

## Method overriding occurs when a child class provides its own implementation of a method already defined in the parent class.
## For ACC-1032, LoyaltyAccount overrides the discount() method and provides a higher discount rate than the parent Account class.
```

```python
#Q33. Use polymorphic payment processors for banking case 1; use sample reference ACC-1033 and explain the result.


class BankTransfer:

  def __init__(self,Account_Number,Amount):
    self.Account_Number = Account_Number
    self.Amount = Amount

  def Process_Payment(self):
    print (f"{self.Account_Number} Bank Transfer payment method {self.Amount}")

class UPI:

  def __init__(self,Account_Number,Amount):
    self.Account_Number = Account_Number
    self.Amount = Amount

  def Process_Payment(self):
    print (f"{self.Account_Number} UPI payment method {self.Amount}")

class NetBanking:

  def __init__(self,Account_Number,Amount):
    self.Account_Number = Account_Number
    self.Amount = Amount

  def Process_Payment(self):
    print (f"{self.Account_Number} Net-Banking Transfer payment method {self.Amount}")

payment = [BankTransfer("ACC-1033",10000),UPI("Acc-1033",8888),NetBanking("Acc-1033",7908)]

for i in payment:
  i.Process_Payment()

### Polymorphism allows different classes to use the same method name while providing different implementations.
### In this example, BankTransfer, UPI, and CardPayment all implement process_payment(), and Python executes the correct version based on the object type.
```
