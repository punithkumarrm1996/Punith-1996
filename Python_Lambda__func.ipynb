```python
#Using reduce(), calculate the sum of all numbers.
from functools import reduce
numbers = [10, 20, 30, 40, 50]
sum=reduce(lambda x,y: x+y,numbers)
print(sum)
```

```python
#Using reduce(), calculate the product of all numbers.
numbers = [2, 3, 4, 5]
product=reduce(lambda x,y:x*y,numbers)
print(product)
```

```python
#Using reduce(), find the maximum number.
numbers = [45, 12, 89, 34, 67, 91, 28]
max=reduce(lambda x,y:x if x>y else y,numbers)
print(max)
```

```python
#Using reduce(), combine all strings into one sentence.
words = ["Python", " ", "is", " ", "powerful"]
sentence=reduce(lambda x,y:x+""+y,words)
print(sentence)
```

```python
#Using reduce(), calculate the total number of characters in all words.
words = ["Python", "Azure", "Cloud", "AI"]
total=reduce(lambda x,y:x+len(y),words,0)
print(total)
```

**Section 5: Combined Lambda, map(), filter(), and reduce() — 20 Questions**

```python
#Using filter() and map():
#Extract all even numbers.
#Calculate the square of every even number.
#Return the squared values as a list.
numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
even_square=list(map(lambda x:x**2,filter(lambda x:x%2==0,numbers)))
print(even_square)
```

```python
#Using filter() and reduce():
#Extract all odd numbers.
#Calculate their total sum.
numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
odd_sum= reduce(lambda x, y: x + y, filter(lambda x: x % 2 != 0, numbers))
print(odd_sum)
```

```python
#Using filter() and map():
#Extract numbers greater than 50.
#Multiply every selected number by 2.
numbers = [10, 25, 40, 55, 70, 85, 100]
great_multi=list(map(lambda x:x*2,filter(lambda x:x>50,numbers)))
print(great_multi)
```

```python
#Using filter(), map(), and reduce():
#Extract positive numbers.
#Calculate the square of every positive number.
#Calculate the total sum of the squared values.
numbers = [-15, 20, -35, 40, 50, -10, 60]
psm=reduce(lambda x,y:x+y,map(lambda x:x**2,filter(lambda x:x>0,numbers)))
print(psm)
```

```python
#Using filter() and map():
#Extract words containing more than five characters.
#Convert the selected words to uppercase.
words = ["python", "azure", "ai", "developer", "cloud", "analytics", "spark"]
upper=list(map(lambda x:x.upper(),filter(lambda x:len(x)>5,words)))
print(upper)
```

```python
#Using filter() and map():
#Extract names beginning with "A".
#Convert the selected names to lowercase.
names = ["Anil", "Rahul", "Ajay", "Meena", "Arjun", "Kiran", "Akhil"]
lower=list(map(lambda x:x.lower(),filter(lambda x:x.startswith("A"),names)))
print(lower)
```

```python
#Using filter() and reduce():
#Extract numbers divisible by both 3 and 5.
#Calculate their total sum.
numbers = [3, 5, 10, 15, 18, 30, 42, 45, 60, 75]
div_sum=reduce(lambda x, y: x + y, filter(lambda x: x % 3 == 0 and x % 5 == 0, numbers))
print(div_sum)
```

```python
#Using filter() and reduce():
#Extract palindrome words.
#Join them into one string separated by " | ".
words = ["madam", "python", "level", "azure", "radar", "cloud", "civic"]
str_palindrome=reduce(lambda x, y: x + " | " + y, filter(lambda x: x == x[::-1], words))
print(str_palindrome)
```

```python
#Using filter(), map(), and reduce():
#Extract temperatures greater than 30°C.
#Convert the selected values into Fahrenheit.
#Calculate the average Fahrenheit temperature.

temperatures = [18, 24, 30, 35, 40, 12, 28, 33, 15, 42]
great=filter(lambda x: x > 30, temperatures)
f=map(lambda x: (x * 9/5) + 32, great)
total=reduce(lambda x, y: x + y, f)
count=len(list(filter(lambda x: x > 30, temperatures)))
average = total / count
print(average)
```

```python
#Using filter() and reduce():
#Extract students who passed.
#Find the topper among the passed students.

students = [
    {"name": "Ravi", "marks": 92, "passed": True},
    {"name": "Kiran", "marks": 68, "passed": True},
    {"name": "Meena", "marks": 85, "passed": True},
    {"name": "Ajay", "marks": 48, "passed": True},
    {"name": "Sneha", "marks": 35, "passed": False},
    {"name": "Rahul", "marks": 95, "passed": True}
]

passed=reduce(lambda x, y: x if x["marks"] > y["marks"] else y,filter(lambda x: x["passed"], students))
print(passed)
```

```python
#Using filter(), map(), and reduce():
#Extract students scoring more than 70.
#Extract only their marks.
#Calculate the average marks.

students = [
    {"name": "Ravi", "marks": 92},
    {"name": "Kiran", "marks": 68},
    {"name": "Meena", "marks": 85},
    {"name": "Ajay", "marks": 48},
    {"name": "Sneha", "marks": 79},
    {"name": "Rahul", "marks": 95}
]

marks=list(map(lambda x: x["marks"],filter(lambda x: x["marks"] > 70, students)))
total = reduce(lambda x, y: x + y, marks)
average = total / len(marks)
print(average)
```

```python
#Using filter(), map(), and reduce():
#Extract products currently in stock.
#Calculate each product's inventory value using price × stock.
#Calculate the total inventory value.

products = [
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0},
    {"name": "SSD", "price": 6500, "stock": 12}
]

inventory_value=reduce(lambda x, y: x + y,map(lambda x: x["price"] * x["stock"],filter(lambda x: x["stock"] > 0, products)))
print(inventory_value)
```

```python
#Using filter(), map(), and reduce():

#Extract products belonging to the Electronics category.
#Apply a 20% discount.
#Calculate the total discounted value.

products = [
    {"name": "Laptop", "price": 55000, "category": "Electronics"},
    {"name": "Chair", "price": 4500, "category": "Furniture"},
    {"name": "Monitor", "price": 18000, "category": "Electronics"},
    {"name": "Desk", "price": 8500, "category": "Furniture"},
    {"name": "SSD", "price": 6500, "category": "Electronics"}
]

discount_value=reduce(lambda x, y: x + y,map(lambda x: x["price"] * 0.80, filter(lambda x: x["category"] == "Electronics", products)))
print(discount_value)
```

```python
#Using filter() and reduce():
#Extract delivered orders.
#Calculate the total delivered revenue.

orders = [
    {"order_id": 1001, "amount": 1200, "status": "Delivered"},
    {"order_id": 1002, "amount": 3500, "status": "Pending"},
    {"order_id": 1003, "amount": 8500, "status": "Delivered"},
    {"order_id": 1004, "amount": 4200, "status": "Cancelled"},
    {"order_id": 1005, "amount": 7600, "status": "Delivered"},
    {"order_id": 1006, "amount": 5100, "status": "Pending"}
]

total_revenue = reduce(lambda x, y: x + y["amount"],filter(lambda x: x["status"] == "Delivered", orders),0)
print(total_revenue)
```

```python
#Using filter(), map(), and reduce():
#Extract transactions whose amount is greater than 5000.
#Add 18% GST to every selected transaction.
#Calculate the grand total after GST.

transactions = [
    {"id": 1, "amount": 1200},
    {"id": 2, "amount": 5200},
    {"id": 3, "amount": 18000},
    {"id": 4, "amount": 2500},
    {"id": 5, "amount": 9600},
    {"id": 6, "amount": 7200}
]

grand_total = reduce(lambda x, y: x + y,map(lambda x: x["amount"] * 1.18, filter(lambda x: x["amount"] > 5000, transactions)))
print(grand_total)
```

```python
#Using filter(), map(), and reduce():
#Extract active salespeople from the South region.
#Calculate a 5% incentive for each selected salesperson.
#Create a list containing salesperson name, sales amount, and incentive.
#Calculate the total incentive payable.

sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]

selected_sales = list(map(lambda x: {"salesperson": x["salesperson"],"sales": x["sales"],"incentive": x["sales"] * 0.05},
                          filter(lambda x: x["region"] == "South" and x["active"], sales_records)))
total_incentive = reduce(lambda x, y: x + y["incentive"],selected_sales,0)

print(selected_sales)
print(total_incentive)
```
