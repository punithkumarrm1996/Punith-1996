Banking Assignment — Regular Expressions

```python
#Q1. Find a account code for banking case 1; use sample reference ACC-1001 and explain the result.


import re

account_number = ["ACC-1001","ABC-1001","ACC1001","ACC-9876"]

pattern = r"^ACC-\d{4}$"

Vaild_Account_Numbers = [i for i in account_number if re.match(pattern,i)]
print("Valid Account Numbers:",Vaild_Account_Numbers)
```

```python
#Q2. Validate a six-digit pin for banking case 1; use sample reference ACC-1002 and explain the result.


import re

Account_number = "ACC-1002"
pin = "123456"

pattern = r"^\d{6}$"

if re.match(pattern,pin):
  print("Account Number:",Account_number)
  print("pin:",pin)
  print("Vaild Pin")
else:
  print("Account Number:",Account_number)
  print("pin:",pin)
  print("Not Vaild pin")
```

```python
#Q31. Validate an account reference with optional category prefix for banking case 1; use sample reference ACC-1032 and explain the result.


import re

account_references = [
    "ACC-1032",
    "SAV-ACC-1032",
    "CUR-ACC-1032",
    "VIP-ACC-1032",
    "SAV-ACC-103",
    "ACC1032"
]

pattern = r"^(?:[A-Z]{3}-)?ACC-\d{4}$"

for account in account_references:
    if re.fullmatch(pattern, account):
        print(account, "-> Valid")
    else:
        print(account, "-> Invalid")
```

```python
#Q32. Normalize account holder phone numbers for banking case 1; use sample reference ACC-1034 and explain the result.


import re

Acc_num = "ACC-1034"

phones = ["9876543210","98765-43210","(987) 654-3210","+91 9876543210","98765 43210","+91-98765-43210"]

for i in phones:
  normalized = re.sub(r"\D","",i)

  if len(normalized) > 10:
    normalized = normalized[-10:]

  print("Account Number:",Acc_num)
  print("Original Phone Number:",i)
  print("Normalized:",normalized)
  print()
```

```python
#Q33. Find repeated account words for banking case 1; use sample reference ACC-1035 and explain the result.


import re

Acc_num = "ACC-1035"

details = """
account account created successfully
customer customer updated
balance verified
loan loan approved
"""

pattern = r"\b(\w+)\s+\1\b"  # \1 - used for Same word captured earlier

matches = re.findall(pattern, details)

print("Account Number:", Acc_num)
print("Repeated Words:", matches)
```

```python
#Q34. Validate fee waiver start and end anchors for banking case 1; use sample reference ACC-1036 and explain the result.


import re

Acc_num = "ACC-1036"

fee_codes = ["FW-1001","FW-2500","FW-9999","FW1001","AB-1234","FW-12345"]

pattern = r"^FW-\d{4}$"

for code in fee_codes:
    if re.fullmatch(pattern, code):
        print("Account Number:", Acc_num)
        print("Valid Fee Waiver Code:", code)
        print()
```

```python
#Q35. Redact emails from support text for banking case 1; use sample reference ACC-1037 and explain the result.


import re

Acc_num = "ACC-1037"

details = """
Customer query received from john@gmail.com
Escalation sent to support@banking.co.in
Backup email: teja.k123@outlook.com
"""

pattern = r"\b[A-Za-z0-9._%+-]+@[a-zA-Z0-9.-]+.[a-zA-Z]{2,}\b"

redact_text = re.sub(pattern,"[REDACT]",details)

print("Account Number:",Acc_num)
print(redact_text)
```

```python
#Q37. Parse inventory key-value pairs for banking case 1; use sample reference ACC-1039 and explain the result.


import re

Acc_num = "ACC-1039"

details = """Cards=150 ChequeBooks=75
Passbooks=200 LoanFiles=50
"""

pattern = r"(\w+)=(\d+)"

matches = re.findall(pattern, details)

print("Account Number:", Acc_num)

for key, value in matches:
    print("Item:", key)
    print("Quantity:", value)
    print()
```

```python
#Q38. Use lookahead for strong fee waiver tokens for banking case 1; use sample reference ACC-1040 and explain the result.


import re

Acc_num = "ACC-1040"

details = """FWA HIGH FWB LOW
FWC HIGH FWD MEDIUM FWE HIGH
"""

pattern = r"\bFW\w+(?=\s+HIGH)"   ## Positive lookahead which means after FW w+ it will see ig HIGH is there or not    re.IGNORECASE is used for case sententive

matches = re.findall(pattern, details)

print("Account Number:", Acc_num)
print("Strong Fee Waiver Tokens:", matches)
```
