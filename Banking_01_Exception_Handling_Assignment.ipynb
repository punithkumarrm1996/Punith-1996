Banking Assignment — Exception Handling

```python
#Q1. Parse a deposit amount for banking case 1; use sample reference ACC-1001 and explain the result.

account_num = "ACC-1001"

try:
    deposit_amount = int(input("Enter the Deposit Amount: "))

    if deposit_amount > 0:
        print(f"Account: {account_num}")
        print(f"Deposit Amount: {deposit_amount}")
    else:
        print("Deposit amount must be greater than 0.")

except ValueError:
    print("Invalid input. Please enter a valid integer amount.")
```

```python
#Q2. Handle a missing account key for banking case 1; use sample reference ACC-1002 and explain the result.

account_details = {
    "account_number" : "ACC-1002",
    "Holder_name" : "Yuvateja"}

try:
  print("Account Number:",{account_details["account_number"]})
  print("Balance:",{account_details["balance"]})


except KeyError:
  print("Missing account details key: 'balance'")
```

```python
#Q3. Prevent division by zero in average balance for banking case 1; use sample reference ACC-1003 and explain the result.

account_number = "ACC-1003"
total_balance = 50000
no_of_accounts = 0

try:
  avg_balance = total_balance/no_of_accounts
  print("Average Balnce",avg_balance)

except ZeroDivisionError:
  print("Account Number ", account_number)
  print("Cannot calculate average balance because the number of accounts is zero.")
```

```python
#Q4. Open a transaction file safely for banking case 1; use sample reference ACC-1004 and explain the result.

account_number = "ACC-1004"

try:
  file = open("Account_Number.txt",'r')
  print("Account Number",account_number)
  print(file.read)

except FileNotFoundError:
  print("Account Number",account_number)
  print("Transaction file not found.")
```

```python
#Q5. Reject a negative withdrawal for banking case 1; use sample reference ACC-1005 and explain the result.


account_number = "ACC-1005"

try:
  withdrawal_amount = int(input("Enter withdraw amount:"))
  if withdrawal_amount < 0:
    raise ValueError("Withdraw amount should not be negative")

  print("Account Number",account_number)
  print("Withdraw Amount", withdrawal_amount)

except ValueError as e:
  print(e)

## Here we are assigning the value error with some text and xalling it with e so that what ever the error msg it throw to input msg.
```

```python
#Q6. Catch invalid tenure input for banking case 1; use sample reference ACC-1006 and explain the result.


account_number = "ACC-1006"
tenture = -10

try:
  if tenture < 0:
    raise ValueError("Tenture should be greater than 0")

  print("Account Number:",account_number)
  print("Tenture:",tenture)

except ValueError as e:
  print("Account Number:",account_number)
  print(e)

##  same as previous example we are raising the error if tenture is invalid
```

```python
#Q7. Use else after pin-format validation for banking case 1; use sample reference ACC-1007 and explain the result.


account_number = "ACC-1007"

pin = input("Enter the pin:")

try:
  if len(pin) != 4:
    raise ValueError("Pin should be 4 digits")

except ValueError as e:
  print("Account number",account_number)
  print(e)

else:
  print("Account number",account_number)
  print("Pin Entered Correctly")

### here lenth of pin not equal to 4 it will throw error and we are calling that error as e, if pin is entered correctly it will dispaly else condition.
```

```python
#Q8. Use finally to close a session for banking case 1; use sample reference ACC-1008 and explain the result.


acc_number = "ACC-1008"

try:
  print("session started")
  print("Account Number", acc_number)
  Balance = 10000
  withdraw = 2000
  if withdraw < 0 and Balance > withdraw:
    raise ValueError("withdraw should be greater than 0 and it should be greater than balance ")
  Remaining_balance = Balance - withdraw
  print("Balance",Remaining_balance)

except ValueError as e:
  print(e)

finally:
  print("Session Closed")
```

```python
#Q31. Create insufficientfundserror for banking case 1; use sample reference ACC-1031 and explain the result.


class InsufficientFundsError(Exception):
  pass
account_number = "ACC-1031"
Balance_amount = 1500
Withdraw_amount = 50000

try:
  if Withdraw_amount > Balance_amount:
    raise InsufficientFundsError('Insufficient Funds')

  Balance_amount -= Withdraw_amount
  print("Remaining Balance:",Balance_amount)

except InsufficientFundsError as e:
  print("Account Balance:",Balance_amount)
  print(e)

### Since there is InsufficientFundsError we are cresting a class for the exception and calling it.
```

```python
#Q32. Chain invalidtransactionerror from valueerror for banking case 1; use sample reference ACC-1032 and explain the result.


class InvalidTransactionError(Exception):
  pass
Account_number = "ACC-1032"
Amount = "YUVA"
try:
  try:
    int(Amount)
  except ValueError as e:
    raise InvalidTransactionError("Invalid Transaction") from e
except InvalidTransactionError as e:
  print('Account Number:',Account_number)
  print(e)
```

```python
#Q40. Collect all kyc validation errors for banking case 1; use sample reference ACC-1040 and explain the result.


account_number = "Acc-1040"
KYC_Details = {"Name":"","PAN":"","DOB":"","Aadhar":""}
errors = []

try:
  for key,value in KYC_Details.items():
    if not value:
      errors.append(f"{key} value Missing in KYC Details")
  if errors:
    raise ValueError("\n".join(errors))

except ValueError as e:
  print("Account Number:",account_number)
  print("KYC Validation Errors:")
  print(e)
```
